# Assignment 4 — Automated Insight Generation
## Auto-Analytics Engine for District Healthcare Performance Data

This notebook demonstrates the complete step-by-step pipeline for:
1. **Part A:** Data Loading, Missing Value Audit & Schema Validation
2. **Part B:** Month-over-Month (MoM) Trend Detection with Configurable Thresholds
3. **Part C:** Outlier Detection using IQR and Z-score Methods
4. **Part D:** Pearson Correlation Matrix & High-Correlation Pair Flagging
5. **Part E:** Dynamic Human-Readable Insight Generation & CSV Export
6. **Part F:** Summary Visualizations & Metrics

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.engine import AutoAnalyticsEngine
from src.exporter import export_insights_to_csv, export_correlation_matrix_to_csv

# Set plot styles
plt.style.use('seaborn-v0_8-whitegrid')
print('✓ Data science environment initialized.')
```

### Part A — Data Loading & Validation
Load the CSV with Pandas and display `head()`, `info()`, and missing-value counts per column.

```python
data_path = 'data/district_healthcare_data.csv'
df = pd.read_csv(data_path)

print('--- DATAFRAME HEAD ---')
display(df.head())

print('\n--- DATAFRAME INFO ---')
df.info()

print('\n--- MISSING VALUE COUNT PER COLUMN ---')
print(df.isnull().sum())
```

### Part B, C, D & E — Execute AutoAnalyticsEngine

```python
engine = AutoAnalyticsEngine(df)

# Execute full engine pipeline with default parameters
results = engine.generate_all_insights(
    trend_threshold_pct=10.0,
    outlier_method='iqr',
    outlier_threshold=1.5,
    correlation_threshold=0.70
)

print(f'Total Insights Generated: {results["total_insights"]}')
print(f'Severity Counts: {results["severity_counts"]}')
print(f'Insight Types: {results["insight_type_counts"]}')
```

### Part E — Generated Insights Table

```python
insights_df = pd.DataFrame(results['insights'])
display(insights_df[['insight_id', 'type', 'indicator', 'entity', 'period', 'severity', 'explanation']])
```

### Part D — Correlation Matrix Visualizations

```python
corr_df = pd.DataFrame(results['correlation_matrix'])
plt.figure(figsize=(8, 6))
sns.heatmap(corr_df, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1, linewidths=1)
plt.title('Pearson Correlation Matrix Across Numerical Healthcare Indicators')
plt.show()
```

### Export Deliverables
Save `insights.csv` and `correlation_matrix.csv` to the output folder.

```python
export_insights_to_csv(results['insights'], 'insights.csv')
export_correlation_matrix_to_csv(results['correlation_matrix'], 'correlation_matrix.csv')
print('✓ Output CSV files generated successfully!')
```
